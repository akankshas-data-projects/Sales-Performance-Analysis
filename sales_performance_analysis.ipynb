In python everything is an object - so when the object data type is object that means it can be any type of data like text, string or any mixed type of data or a string

```python
#To convert object data type of Order date into datetime format

data["Order Date"] = pd.to_datetime(data["Order Date"])
```

```python
data.info()
```

```python
#To convert ship date column to datetime format

data["Ship Date"] = pd.to_datetime(data["Ship Date"])
```

```python
data.info()
```

```python
data.head()
```

```python
#Adding three more columns
#Monthly order analysis 
data["Order Month"] = data["Order Date"].dt.month
```

```python
data.info()
```

```python
#Adding a year column for yearly analysis

data["Order Year"] = data["Order Date"].dt.year
```

```python
data.info()
```

```python
#Adding a column to see if the order reaches in the same week or no

data["Order Day of week"] = data["Order Date"].dt.dayofweek
```

```python
data.info()
```

```python
data.head()
```

```python
#Monthly sales analysis

sales_by_month = data.groupby('Order Month')['Sales'].sum().reset_index()
```

```python
#printing the sales_by_month 
sales_by_month
```

```python
#Plotting the data in the graph format

import matplotlib.pyplot as plt
plt.figure(figsize = (8,8))
plt.plot(sales_by_month['Order Month'], sales_by_month['Sales'], marker = '.')
plt.xlabel('Order Month')
plt.ylabel('Total Sales')
plt.title('Monthly Sales Trend')
plt.show()
```

```python
data.head()
```

```python
#Sales by category

sales_by_category = data.groupby('Category')['Sales'].sum().reset_index()
```

```python
sales_by_category
```

```python
#Making a pie chart:
import matplotlib.pyplot as plt
plt.figure(figsize =(5, 5))
plt.pie(sales_by_category['Sales'],
        labels = sales_by_category['Category'],
        autopct = '%1.1f%%',
        startangle = 90)
plt.title('Sales Share by Catgeory')
plt.show()
```
